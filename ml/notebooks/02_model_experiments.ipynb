# 02. Baseline Model Experiments
## Subject-Level Multi-Model Comparison

This notebook evaluates baseline classifiers (Logistic Regression, K-Nearest Neighbors, Random Forest, and Support Vector Machine) using subject-level stratified group cross-validation.


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, "..")
from src.features.preprocessing import load_dataset, get_X_y_groups
from src.data.splitting import subject_level_train_test_split
from src.models.baseline import run_baseline_comparison

%matplotlib inline
sns.set_theme(style="whitegrid")
```

### 1. Load Data and Perform Subject-Level Splitting


```python
df = load_dataset()
df_train, df_test = subject_level_train_test_split(df)
print(f"Train samples: {len(df_train)}, Test samples: {len(df_test)}")
assert set(df_train["subject_id"]).isdisjoint(set(df_test["subject_id"])), "Leakage detected!"
```

### 2. Run Baseline Models Comparison


```python
baseline_df = run_baseline_comparison(df_train, df_test)
baseline_df
```

### 3. Visualize Baseline Performance


```python
plt.figure(figsize=(8, 5))
df_plot = baseline_df.melt(id_vars=["label"], value_vars=["accuracy", "f1_macro", "f1_weighted"], var_name="Metric", value_name="Score")
sns.barplot(data=df_plot, x="label", y="Score", hue="Metric", palette="Blues_d")
plt.title("Baseline Model Comparison on Test Set", fontsize=13)
plt.ylim(0, 1.0)
plt.xlabel("Model")
plt.ylabel("Score")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```
